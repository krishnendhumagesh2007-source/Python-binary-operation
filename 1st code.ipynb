```python
!pip install lxml
import pandas as pd

# Employees dataset (Excel)
employees = pd.DataFrame({
    "employee_id": [101, 102, 103, 104],
    "name": ["Alice", "Bob", "Charlie", "David"],
    "department_id": [1, 2, 1, 3]
})
employees.to_excel("employees.xlsx", index=False)

# Departments dataset (XML)
departments = pd.DataFrame({
    "department_id": [1, 2, 3],
    "department_name": ["HR", "Finance", "IT"]
})
departments.to_xml("departments.xml", index=False)

# Salaries dataset (JSON)
salaries = pd.DataFrame({
    "employee_id": [101, 102, 103, 104],
    "salary": [50000, 60000, 55000, 65000]
})
salaries.to_json("salaries.json", orient="records", lines=True)
```

```python
import pandas as pd

# Read Excel
employees_df = pd.read_excel("employees.xlsx")

# Read XML
departments_df = pd.read_xml("departments.xml")

# Read JSON
salaries_df = pd.read_json("salaries.json", lines=True)

print("Employees Data:\n", employees_df)
print("\nDepartments Data:\n", departments_df)
print("\nSalaries Data:\n", salaries_df)
```

```python
# Merge employees with departments (on department_id)
merged_df = pd.merge(employees_df, departments_df, on="department_id", how="inner")

# Merge with salaries (on employee_id)
merged_df = pd.merge(merged_df, salaries_df, on="employee_id", how="inner")

print("Merged Data:\n", merged_df)
```

```python
# Save to CSV (change path as needed)
merged_df.to_csv(r"C:\Users\krish\OneDrive\Documents\merged.csv", index=False)

print("Merged data saved successfully!")
```
